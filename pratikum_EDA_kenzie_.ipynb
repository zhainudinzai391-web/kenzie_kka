```python
import numpy as np
nilai = np.array([80, 75, 90, 60, 88])
print(nilai.mean()) # rata-rata
print(nilai.max()) # nilai tertinggi
print(nilai * 2) # operasi langsung ke semua elemen (vectorized)
import pandas as pd
data = {
'nama': ['Andi', 'Budi', 'Citra'],
'nilai': [80, 75, 90]
}
df = pd.DataFrame(data)
print(df)
```

Hasil: NumPy array berhasil menghitung harga diskon ([4500., 6300., ...]), sedangkan List Python menghasilkan TypeError.

Penyebab:

NumPy Array menggunakan konsep vektorisasi, di mana operator * langsung mengalikan angka 0.9 ke setiap elemen di dalamnya secara matematis.

List Python mengartikan operator * sebagai pengulangan data (bukan operasi matematika). Karena pengulangan hanya bisa dilakukan dengan bilangan bulat (integer), mengalikan list dengan desimal (0.9) menyebabkan eror.

Kolom dengan Data Kosong: menu (baris ke-4) dan terjual (baris ke-2).

Risiko Tanpa Pembersihan Data:

Statistik Tidak Akurat: Perkalian harga dan unit terjual akan menghasilkan NaN (kosong), sehingga total pendapatan dan rata-rata penjualan menjadi salah atau bias.

Pemasukan Tidak Teridentifikasi: Terjadi transaksi penjualan (18 unit, Rp8.000), tetapi tidak diketahui menu apa yang dijual.

Laporan Bisnis Menyesatkan: Pengambilan keputusan (seperti menentukan menu terlaris) menjadi tidak valid karena datanya cacat dan belum terkelompokkan dengan benar.

```python
import pandas as pd

# Definisikan ulang data_kantin dan DataFrame agar kode dapat berjalan mandiri
data_kantin = {
    'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
    'harga': [12000, 4000, 10000, 4000, 8000],
    'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)

# Menyimpan ke CSV dan membaca kembali
df.to_csv('data_kantin.csv', index=False)
df = pd.read_csv('data_kantin.csv')

print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
```

Kolom: menu dan terjual (memiliki 4 data non-null dari total 5 baris).

Artinya:

Ada Data Kosong (Missing Values): Terdapat nilai NaN/None pada kolom tersebut sebanyak selisihnya (1 data kosong di masing-masing kolom).

Perlu Pembersihan Data: Menandakan bahwa dataset belum utuh dan perlu diperbaiki (diisi nilainya atau dihapus) sebelum dianalisis lebih lanjut agar tidak menyebabkan eror atau bias statistik.

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

fillna(0) pada kolom terjual:

Alasan: Asumsi paling logis untuk data penjualan yang kosong adalah belum ada yang terjual (0 unit).

Tujuan: Mempertahankan baris data tersebut agar perhitungan matematika (seperti total omzet) tetap berjalan akurat tanpa eror NaN.

dropna() pada kolom menu:

Alasan: Kolom menu adalah identitas utama produk. Data harga dan unit terjual tidak memiliki arti bisnis jika tidak diketahui produk apa yang dijual.

Tujuan: Nama produk tidak bisa ditebak secara akurat, sehingga menghapus baris cacat ini jauh lebih aman untuk menjaga keakuratan laporan.

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

Jumlah Baris: Tetap 4 baris (tidak berkurang), karena tidak ada baris yang 100% identik di seluruh kolom (transaksi Es Teh memiliki nilai terjual yang berbeda, yaitu 40 dan 35 unit).

Pentingnya Tipe Data (dtypes):

Mencegah Eror Kalkulasi: Angka berformat teks (string) tidak bisa dihitung secara matematis (contoh: '4000' + '4000' menjadi '40004000', bukan 8000).

Efisiensi & Kompatibilitas: Memastikan penggunaan memori lebih hemat serta memungkinkan fungsi statistik (sum, mean) dan visualisasi grafik berjalan dengan benar.

```python
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting
df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan
ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```

Menu Tertinggi: Es Teh (Total Pendapatan: Rp300.000).

Manfaat Pengambilan Keputusan Bisnis:

Manajemen Stok: Memastikan pasokan bahan baku Es Teh selalu aman karena merupakan sumber utama omzet kantin.

Strategi Bundling: Membuat paket hemat (misal: Mie Ayam + Es Teh) untuk mendongkrak penjualan menu lain yang volume penjualannya rendah.

Evaluasi Menu: Menjadi dasar untuk mengevaluasi, merestrukturisasi harga, atau mengganti menu yang pendapatannya rendah/nol (seperti Mie Ayam).
